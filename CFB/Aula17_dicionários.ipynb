```python
Carro={
    "Fabricante":"Valor",
    "Modelo":"HRV",
    "Ano":"2016",
    "Cor":"Prata",
    "Câmbio":"Automático"
}
```

```python
Carro.pop("Fabricante")
#Carro.clear() reseta o dicionário

print(f'Tamanho do dicionário: {len(Carro)}')
```

```python
for c, v in Carro.items():
    print(f"{c}: {v}")
```

```python
carros={
    "carro1":{"Fabricante":"Honda","Modelo":"HRV"},
    "carro2":{"Fabricante":"Volksvagem","Modelo":"Golf"},
    "carro3":{"Fabricante":"Ford","Modelo":"Focus"}
}
```

```python
print(carros['carro1']['Fabricante'])
```

```python
for c1,v1 in carros.items():
  for c2,v2 in v1.items():
    print(f'{c2}:{v2}')
```
